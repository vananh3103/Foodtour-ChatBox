```python
%pip install pandas openpyxl

from pathlib import Path
import json
import pandas as pd

# Khởi tạo đường dẫn dữ liệu
DATA_DIR = Path("dataset")
TRAIN_FILE = DATA_DIR / "train_vietnam_tourism.json"
VALID_FILE = DATA_DIR / "valid_vietnam_tourism.json"

# Đọc dữ liệu JSON
with TRAIN_FILE.open("r", encoding="utf-8") as file:
    train_data = json.load(file)

with VALID_FILE.open("r", encoding="utf-8") as file:
    valid_data = json.load(file)

# Đọc tất cả file Excel trong thư mục dataset
excel_data = {
    excel_file.stem: pd.read_excel(excel_file)
    for excel_file in DATA_DIR.glob("*.xlsx")
}

# Kiểm tra kết quả đọc JSON
train_records = train_data.get("data", [])
valid_records = valid_data.get("data", [])
print(f"Train JSON: đọc được {len(train_records)} mẫu")
print(f"Valid JSON: đọc được {len(valid_records)} mẫu")
if train_records:
    print("Mẫu train đầu tiên:", train_records[0].get("title", "Không có title"))

# Kiểm tra kết quả đọc Excel và xem trước 3 dòng
if excel_data:
    print(f"Đọc được {len(excel_data)} file Excel:")
    for file_name, dataframe in excel_data.items():
        print(f"\\n{file_name}: {dataframe.shape[0]} dòng, {dataframe.shape[1]} cột")
        display(dataframe.head(3))
else:
    print(f"Không tìm thấy file .xlsx trong thư mục: {DATA_DIR.resolve()}")
```
